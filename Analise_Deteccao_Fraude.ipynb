```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# --- PASSO 1: CRIAR A BASE DE DADOS (Simulando o Banco) ---
np.random.seed(42)
n_transacoes = 1000
data = {
    'Valor': np.random.exponential(scale=100, size=n_transacoes),
    'Localidade': np.random.choice(['Online', 'Loja Física', 'Internacional'], n_transacoes),
    'Hora_Dia': np.random.randint(0, 24, n_transacoes),
    'Fraude': np.random.choice([0, 1], size=n_transacoes, p=[0.97, 0.03])
}
df_fraude = pd.DataFrame(data)

# --- PASSO 2: PREPARAÇÃO (TRADUÇÃO) ---
le = LabelEncoder()
df_fraude['Local_N'] = le.fit_transform(df_fraude['Localidade'])

# --- PASSO 3: MACHINE LEARNING (O DETETIVE) ---
X = df_fraude[['Valor', 'Hora_Dia', 'Local_N']]
y = df_fraude['Fraude']

# Dividindo 70% para treino e 30% para teste
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Treinando o modelo
modelo_fraude = RandomForestClassifier(n_estimators=100, random_state=42)
modelo_fraude.fit(X_train, y_train)

# --- AJUSTANDO O ROBÔ PARA DAR PESO ÀS FRAUDES ---

# No passo 4 do código anterior, mude apenas a linha do modelo para esta:
modelo_fraude_ajustado = RandomForestClassifier(n_estimators=100, class_weight='balanced', random_state=42)
modelo_fraude_ajustado.fit(X_train, y_train)

# Gerando novas previsões
previsoes_novas = modelo_fraude_ajustado.predict(X_test)

print("--- RELATÓRIO DO DETETIVE IA (AGORA COM PESO BALANCEADO) ---")
print(classification_report(y_test, previsoes_novas))
```

```python
# --- FORÇANDO UM PADRÃO PARA O APRENDIZADO ---

# Vamos dizer que toda fraude (1) tem um valor adicionado de 500 reais
# Isso cria uma "pista" para o nosso detetive IA
df_fraude.loc[df_fraude['Fraude'] == 1, 'Valor'] += 500

# Agora, vamos treinar novamente com o peso balanceado
X = df_fraude[['Valor', 'Hora_Dia', 'Local_N']]
y = df_fraude['Fraude']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

modelo_detetive = RandomForestClassifier(n_estimators=100, class_weight='balanced', random_state=42)
modelo_detetive.fit(X_train, y_train)

previsoes_finais = modelo_detetive.predict(X_test)

print("--- RELATÓRIO DO DETETIVE IA (COM PISTA DE VALOR ALTO) ---")
print(classification_report(y_test, previsoes_finais))
```

```python
# 1. Calculando a média primeiro (para o Python não esquecer)
analise_valor = df_fraude.groupby('Fraude')['Valor'].mean()

# 2. Criando o visual do gráfico
plt.figure(figsize=(8,5))
sns.barplot(x=analise_valor.index, y=analise_valor.values, palette='Reds')

plt.title('Valor Médio: Transação Normal (0) vs Fraude (1)')
plt.xticks([0, 1], ['Normal', 'Fraude'])
plt.ylabel('Valor Médio (R$)')
plt.xlabel('Tipo de Transação')

# 3. SALVANDO E MOSTRANDO (Nesta ordem!)
plt.savefig('grafico_fraude.png') 
plt.show()

print("Sucesso! O arquivo 'grafico_fraude.png' foi salvo na sua pasta.")
```
